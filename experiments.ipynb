```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
data = pd.read_csv("Churn_Modelling.csv")
data.head()
```

```python
data.drop(['RowNumber','CustomerId','Surname'],axis = 1,inplace = True)
```

```python
data.head()
```

```python
encoder_gender = LabelEncoder()

data['Gender'] = encoder_gender.fit_transform(data['Gender'])
data
```

```python
from sklearn.preprocessing import OneHotEncoder

encode_geography = OneHotEncoder()

var = encode_geography.fit_transform(data[['Geography']]).toarray()
var
```

```python
geo_df = pd.DataFrame(var,columns = encode_geography.get_feature_names_out(['Geography']))
geo_df
```

```python
data = pd.concat([geo_df,data],axis=1)
```

```python
data.head()
```

```python
data.drop(['Geography'],axis=1,inplace=True)
```

```python
data.head()
```

```python
data.info()
```

```python
## Save encoders and scaler

with open('Label_encoder_gender.pkl','wb') as file:
    pickle.dump(encoder_gender,file)
```

```python
with open('OneHotEncoder_geography.pkl','wb') as file:
    pickle.dump(encode_geography,file)
```

```python
X=data.drop(['Exited'],axis=1)
y = data['Exited']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.1,random_state=42)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

# ANN

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
X_train.shape,X_train.shape[1]
```

```python
model = Sequential([
    Dense(64,activation = 'relu',input_shape = (X_train.shape[1],)), #Hideen Layer 1
    Dense(32,activation = 'relu'), #HL2
    Dense(1,activation = 'sigmoid')
    
])
```

```python
model.summary()
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate=0.01)
loss = tensorflow.keras.losses.BinaryCrossentropy()
```

```python
## compile model

model.compile(optimizer ="adam",loss="binary_crossentropy",metrics=['accuracy'])
```

```python
log_dir = "logs/fit" + datetime.datetime.now().strftime("%Y%m%d - %H%M%S")
tensorflow_callback = TensorBoard(log_dir = log_dir,histogram_freq=1)
```

```python
## Early stopping (epochs control)
early_stopping_callback =EarlyStopping(monitor = 'val_loss',patience = 5,restore_best_weights=True)

```

```python
##Train model
history = model.fit(
    X_train,y_train,validation_data = (X_test,y_test),
    epochs = 100,
    callbacks = [tensorflow_callback,early_stopping_callback]
)
```

```python
model.save('model.h5')
```
